```python
# Day 2 - Data Cleaning
```

```python
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
df["rating"] = df["rating"].fillna("Unknown")
df["date_added"] = df["date_added"].fillna("Unknown")
df["duration"] = df["duration"].fillna("Unknown")
```

```python
# Day 5 - Director Analysis
```

```python
top_directors = df[df["director"] != "Unknown"]["director"].value_counts().head(10)
top_directors
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
top_directors.plot(kind="bar")

plt.title("Top 10 Directors on Netflix")
plt.xlabel("Director")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.show()
```

## Observation

- A few directors have directed multiple titles on Netflix.
- Missing director names were excluded from this analysis.

```python
# Day 5 - Year-wise Content Trend
```

```python
year_counts = df["release_year"].value_counts().sort_index()
year_counts.tail(20)
```

```python
plt.figure(figsize=(12,6))

plt.plot(year_counts.index, year_counts.values, marker="o")

plt.title("Netflix Titles Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

## Observation

- The number of Netflix titles increased rapidly after 2015.
- Recent years contain the highest number of releases.

```python
# Day 6 - Movie Duration Analysis
```

```python
movie_duration = df[df["type"] == "Movie"]["duration"]

movie_duration.value_counts().head(10)
```

```python
movie_duration = df[df["type"] == "Movie"]["duration"]

movie_duration.value_counts().head(10)
```

```python
plt.figure(figsize=(10,5))

movie_duration.value_counts().head(10).plot(kind="bar", color="orange")

plt.title("Top 10 Movie Durations")
plt.xlabel("Duration")
plt.ylabel("Number of Movies")

plt.xticks(rotation=45)

plt.show()
```

## Observation

- Most Netflix movies have durations between 90 and 100 minutes.
- Very short and very long movies are less common.

```python
# Day 6 - Movies and TV Shows by Country
```

```python
country_type = df.groupby("country")["type"].value_counts().unstack().fillna(0)

country_type.head()
```

```python
top_country = country_type.sort_values(by="Movie", ascending=False).head(10)

top_country
```

```python
top_country.plot(kind="bar", figsize=(12,6))

plt.title("Movies and TV Shows by Country")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

## Observation

- The United States has the highest number of Movies.
- India also contributes a large number of Movies.
- TV Shows are fewer than Movies in most countries.

```python
# Day 7 - Monthly Content Added Analysis
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

df["month_added"] = df["date_added"].dt.month_name()

month_counts = df["month_added"].value_counts()

month_counts
```

```python
plt.figure(figsize=(10,5))

month_counts.plot(kind="bar", color="skyblue")

plt.title("Netflix Content Added by Month")
plt.xlabel("Month")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

## Observation

- Some months have significantly more content additions.
- Netflix appears to add content in batches during certain months.

```python
type_counts = df["type"].value_counts()

type_counts
```

```python
plt.figure(figsize=(6,6))

plt.pie(
    type_counts,
    labels=type_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Movies vs TV Shows")

plt.show()
```

## Observation

- Movies occupy the majority of Netflix content.
- TV Shows represent a smaller percentage.

```python
# Day 8 - Top 10 Genres
```

```python
genre = df["listed_in"].str.split(", ").explode()

top_genres = genre.value_counts().head(10)

top_genres
```

```python
plt.figure(figsize=(10,5))

top_genres.plot(kind="bar", color="green")

plt.title("Top 10 Netflix Genres")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")

plt.xticks(rotation=45)

plt.show()
```

## Observation

- Drama and International content are among the most common genres.
- Comedy is also widely represented.

```python
# Day 9 - Longest Movies Analysis
```

```python
movies = df[df["type"] == "Movie"].copy()

movies["minutes"] = movies["duration"].str.replace(" min", "").astype(float)

top_movies = movies.sort_values(by="minutes", ascending=False)

top_movies[["title", "minutes"]].head(10)
```

```python
plt.figure(figsize=(10,5))

plt.bar(top_movies["title"].head(10),
        top_movies["minutes"].head(10),
        color="purple")

plt.xticks(rotation=90)

plt.title("Top 10 Longest Movies")
plt.xlabel("Movie")
plt.ylabel("Duration (Minutes)")

plt.show()
```

```python
tv = df[df["type"] == "TV Show"].copy()

tv["seasons"] = tv["duration"].str.extract("(\d+)").astype(float)

tv["seasons"].describe()
```

```python
plt.figure(figsize=(8,5))

plt.hist(tv["seasons"], bins=10)

plt.title("Distribution of TV Show Seasons")
plt.xlabel("Number of Seasons")
plt.ylabel("Count")

plt.show()
```

```python
year_type = df.groupby(["release_year", "type"]).size().unstack(fill_value=0)

year_type.tail()
```

```python
year_type.plot(figsize=(12,6))

plt.title("Movies vs TV Shows by Release Year")

plt.show()
```

```python
director = df["director"].dropna()

director = director.str.split(", ").explode()

top_directors = director.value_counts().head(10)

top_directors
```

```python
plt.figure(figsize=(10,5))

top_directors.plot(kind="bar", color="red")

plt.title("Top 10 Directors")

plt.show()
```

```python
rating_type = pd.crosstab(df["rating"], df["type"])

rating_type
```

```python
rating_type.plot(kind="bar", figsize=(10,6))

plt.title("Ratings by Content Type")

plt.show()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
```

```python
df["year_added"] = df["date_added"].dt.year
```

```python
year_added = df["year_added"].value_counts().sort_index()

year_added
```

```python
plt.figure(figsize=(12,6))

plt.plot(year_added.index, year_added.values, marker="o")

plt.title("Netflix Titles Added by Year")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")
plt.grid(True)

plt.show()
```

## Observation

- Netflix added very few titles before 2015.
- Content additions increased rapidly after 2016.
- The highest number of titles were added around 2019.

```python
movies = df[df["type"] == "Movie"].copy()

movies["minutes"] = movies["duration"].str.extract("(\d+)").astype(float)

movies[["release_year", "minutes"]].corr()
```

```python
plt.figure(figsize=(5,4))

sns.heatmap(
    movies[["release_year", "minutes"]].corr(),
    annot=True,
    cmap="Blues"
)

plt.show()
```
